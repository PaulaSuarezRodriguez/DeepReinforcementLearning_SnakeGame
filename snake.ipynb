```python
from stable_baselines3.common.env_util import make_vec_env
import tcs

vec_env = make_vec_env(
    "tcs-v2",
    env_kwargs={
        "height": 32,
        "width": 32,
        "render_mode": "human",
        "mp4": True,
        "image_size": 100,
    },
    n_envs=1,
    seed=42,
)

obs = vec_env.reset()

while True:
    actions = [vec_env.action_space.sample()]  # n_envs=1, so 1 action
    obs, rewards, dones, infos = vec_env.step(actions)

    if dones[0]:  # episode finished
        print(infos[0])  # print final info (reward, length, reason)
        break

vec_env.close()
```
